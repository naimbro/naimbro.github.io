## Parte 3 — Columnas: un número por año

En la clase 7 calculaste el porcentaje que ve la economía del país «Mala» o «Muy
mala», año por año. Aquí está la receta, guardada con el nombre `econ`:

```r
econ <- cep %>%
  mutate(econ_mala = sit_econ_pais == "Mala" | sit_econ_pais == "Muy mala") %>%
  group_by(anio) %>%
  summarise(personas = n(),
            pct_econ_mala = mean(econ_mala) * 100)

econ
```

```r
anios <- cep %>%
  mutate(econ_mala = sit_econ_pais == "Mala" | sit_econ_pais == "Muy mala",
         aprueba = aprueba_presidente == "Aprueba") %>%
  group_by(anio) %>%
  summarise(pct_econ_mala = mean(econ_mala) * 100,
            pct_aprueba = mean(aprueba) * 100)

anios
```
