## Paso 3 — La receta

Ésta es la cadena completa, con el ejemplo que vimos en clases: el porcentaje que
menciona la delincuencia, año por año.

```r
cep %>%
  mutate(delinc = problema_1 == "Delincuencia, asaltos y robos") %>%
  group_by(anio) %>%
  summarise(personas = n(),
            pct_delincuencia = mean(delinc) * 100) %>%
  arrange(desc(pct_delincuencia))
```

Cuatro líneas y cada una hace una cosa:

| Línea | Qué hace |
|---|---|
| `mutate()` | crea una columna que vale `TRUE` cuando la respuesta es la que te interesa |
| `group_by()` | parte la tabla en grupos |
| `summarise()` | cuenta cuántos son (`n()`) y promedia los `TRUE`, que es la proporción |
| `arrange()` | ordena el resultado |

**Cómo adaptarla a su encargo**, cambiando una línea a la vez:

- **Otra comparación**: cambien `anio` en el `group_by()` por `gse`, `region` o
  `sexo`.
- **Otra pregunta**: cambien la condición del `mutate()`. Para el encargo 4, por
  ejemplo, es `aprueba_presidente == "Aprueba"`.
- **Sólo algunos años**: agreguen un `filter()` como primera línea, por ejemplo
  `filter(anio >= 2022)` o `filter(anio == 2026)`.
