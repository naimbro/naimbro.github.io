## Parte 4 · Tu primera pregunta: ¿iPhone o Android?

Acá está **la función de hoy**. Se llama `count()` y se lee tal cual suena:
*cuenta, en la base `curso`, cuántos hay de cada `sistema_operativo`*.

````r
count(curso, sistema_operativo)
````

Lee el resultado con calma, porque **todas las respuestas de hoy se ven igual**:

- Una **fila por cada categoría** distinta que existe en esa columna.
- Una columna llamada **`n`** con **cuántas personas** hay en cada una.

Así que la respuesta es: **27 con iPhone, 5 con Android y 1 con otro sistema.**

## Parte 9 · El conteo que decide sus grupos

Ahora el conteo más importante del día: **qué dominio eligió cada uno para su
proyecto del semestre**.

````r
count(curso, dominio, sort = TRUE)
````

Ahí está el curso entero. Dos cosas que vale la pena mirar:

**Salen 10 categorías, pero el curso no eligió 10 temas.** Alguien escribió
`politica` en minúscula, alguien escribió `Medioambiente` en una palabra, y
alguien no eligió del menú y escribió `tecnologia y salud`. Decidir qué hacer con
esos tres **no lo puede hacer R sola**: es una decisión de ustedes, y hay que
dejarla escrita.

**Varias categorías tienen `n = 1`.** Una persona. Si alguien dijera *"el 100% de
los que eligieron transporte piensa X"*, ese 100% **es una sola persona**, y no
describe a nadie más que a ella.

> Ésa es la razón por la que el proyecto del semestre no se puede hacer con esta
> base: **somos 33**. Para hablar de un tema de verdad hay que salir a buscar
> datos que ya existen, levantados por alguien más, sobre miles de personas.
> De eso se trata la segunda mitad de la clase de hoy.

### Las tres ideas

1. **Una función bien entendida rinde más que diez a medias.** Con `count()` sola
   describiste un curso completo.
2. **Un número sin saber sobre cuánta gente está calculado puede mentir.** Un
   `n = 1` no describe a nadie más que a esa persona.
3. **Los datos de verdad vienen sucios**: `3` y `3 horas`, `Las condes` y
   `las condes`. Notarlo ya es parte del trabajo.
