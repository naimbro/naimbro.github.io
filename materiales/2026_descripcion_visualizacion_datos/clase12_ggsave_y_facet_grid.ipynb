## Parte 0 · Punto de partida

```r
library(dplyr)
library(ggplot2)
options(repr.plot.width = 9, repr.plot.height = 5)

problemas <- read.csv("https://raw.githubusercontent.com/naimbro/naimbro.github.io/main/materiales/2026_descripcion_visualizacion_datos/datos/cep_problemas_por_anio.csv")
cep <- read.csv("https://raw.githubusercontent.com/naimbro/naimbro.github.io/main/materiales/2026_descripcion_visualizacion_datos/datos/cep_consolidada_1994_2026.csv")
```

## Parte 1 · `ggsave()`: el gráfico sale de Colab

`ggsave()` guarda **el último gráfico** que dibujaron, con el nombre y el tamaño que le
pidan (en pulgadas).

Para bajarlo a su computador: **panel de archivos** (el ícono de carpeta, a la izquierda) →
clic derecho sobre el PNG → **Descargar**. Ojo: si cierran Colab, el archivo se borra.

```r
delinc <- problemas %>%
  filter(problema == "Delincuencia, asaltos y robos")

ggplot(delinc, aes(x = anio, y = porcentaje)) +
  geom_line()

ggsave("delincuencia.png", width = 8, height = 5)
```

## Parte 2 · `facet_grid()`: cruzar dos variables

`facet_wrap(~ x)` hace un panel por grupo. `facet_grid(filas ~ columnas)` cruza **dos**
variables: aquí, la zona en las filas y el sexo en las columnas. Primero, una tabla con el
porcentaje que nombra la delincuencia como primer problema, por año, sexo y zona.

```r
delinc_grupos <- cep %>%
  mutate(delinc = problema_1 == "Delincuencia, asaltos y robos") %>%
  group_by(anio, sexo, zona) %>%
  summarise(pct = mean(delinc) * 100)

ggplot(delinc_grupos, aes(x = anio, y = pct)) +
  geom_line() +
  facet_grid(zona ~ sexo)
```

## Parte 3 · El título es el hallazgo

La regla de la portada:

- **title:** lo que encontraron, con un número y una comparación.
- **subtitle:** qué se mide.
- **caption:** la fuente.

```r
ggplot(delinc_grupos, aes(x = anio, y = pct)) +
  geom_line() +
  facet_grid(zona ~ sexo) +
  labs(
    title = "Campo: de 13% a 32%. Ciudad: de 20% a 31%",
    subtitle = "% que nombra la delincuencia primero",
    x = NULL, y = NULL,
    caption = "Fuente: CEP 1994-2026, sin 2020")
```
