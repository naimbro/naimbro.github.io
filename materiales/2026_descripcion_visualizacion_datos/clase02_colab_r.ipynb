## Parte 2 · Guardar cosas: los objetos

Hasta ahora R calculaba y olvidaba. Para **guardar** un resultado y reutilizarlo le ponemos un
nombre. A eso le llamamos crear un **objeto** (o una *variable*).

La flecha `<-` significa *"guarda esto adentro de"*. Se lee de derecha a izquierda:

```
mi_edad  <-  19
   ↑         ↑
 nombre    valor
```

> ⌨️ **Atajo:** en Colab puedes escribir la flecha con **Alt + -** (o **Option + -** en Mac).
> El signo `=` también funciona, pero en R la convención es usar `<-`.

````r
mi_edad <- 19        # crea el objeto "mi_edad" y guarda el número 19
mi_edad              # escribir el nombre del objeto muestra su contenido
````

Una vez creado, el objeto se puede usar en cálculos como si fuera el valor que contiene.

````r
mi_edad + 1          # ¿qué edad voy a tener el próximo año?

edad_en_dias <- mi_edad * 365
edad_en_dias
````

### Reglas para nombrar objetos

- Sin espacios: usa `guion_bajo` (`horas_sueno`, `edad_promedio`).
- No empiezan con número: `2edad` da error, `edad2` está bien.
- **R distingue mayúsculas de minúsculas**: `Edad` y `edad` son dos objetos distintos.
- Elige nombres que se entiendan solos: `estatura_cm` es mucho mejor que `x1`.

> 🧠 Esta última regla parece trivial y es de las más importantes del semestre.

## Parte 3 · Tipos de datos: no todo es un número

R distingue distintos **tipos** de información. Los cuatro que usaremos todo el semestre son:

| Tipo en R | Qué es | Ejemplo | En el curso lo llamaremos |
|---|---|---|---|
| `numeric` | números con decimales | `1.75`, `7.5` | variable numérica |
| `integer` | números enteros | `19`, `40` | variable numérica |
| `character` | texto, **siempre entre comillas** | `"Ñuñoa"`, `"Metro"` | variable categórica |
| `logical` | verdadero o falso | `TRUE`, `FALSE` | variable lógica |

La función `class()` nos dice de qué tipo es un objeto.

````r
estatura <- 1.72               # numeric
comuna   <- "Ñuñoa"            # character (¡con comillas!)
usa_metro <- TRUE              # logical (TRUE y FALSE van SIN comillas y en mayúsculas)

class(estatura)
class(comuna)
class(usa_metro)
````

Esta distinción no es un capricho técnico: **determina qué preguntas puedes hacerle a una variable**.

- A una variable numérica le puedes preguntar su **promedio** (¿cuál es la estatura promedio del curso?).
- A una variable categórica le puedes preguntar su **frecuencia** (¿cuántos vienen en metro?).
- Preguntarle el promedio a una comuna no tiene sentido, y R te lo va a hacer notar.

Veamos qué pasa cuando uno se equivoca de tipo:

````r
# Esto va a dar un ERROR a propósito. Ejecútalo y lee el mensaje con calma.
"5" + 3
````

Ese mensaje —`non-numeric argument to binary operator`— significa: *"me pediste sumar algo que no es

## Parte 4 · Vectores: varios datos en un solo objeto

Un curso no tiene una edad: tiene 40. Para guardar varios valores del mismo tipo en un solo objeto
usamos un **vector**, que se construye con la función `c()` (de *combinar*), separando los valores
con comas.

````r
# Las edades de siete compañeros
edades <- c(18, 19, 18, 21, 20, 19, 18)
edades

length(edades)     # ¿cuántos valores tiene el vector?
````

Lo potente de los vectores es que las operaciones se aplican **a todos los elementos a la vez**.

````r
edades + 1         # suma 1 a cada una de las siete edades
````

Y con vectores ya podemos calcular las medidas que vimos en clases. Estas funciones son el corazón
de la estadística descriptiva:

````r
mean(edades)       # media (promedio)
median(edades)     # mediana (el valor del medio)
min(edades)        # mínimo
max(edades)        # máximo
sd(edades)         # desviación estándar (cuánto se dispersan los datos)
````

### La moda

R no trae una función para la moda (el valor más frecuente), así que la construimos en dos pasos.
`table()` cuenta cuántas veces aparece cada valor:

````r
table(edades)                       # tabla de frecuencias

names(which.max(table(edades)))     # el valor que más se repite = la moda
````

Los vectores también sirven para texto:

````r
transporte <- c("Metro", "Micro", "Metro", "Auto", "Metro", "Caminando", "Micro")

table(transporte)                   # frecuencia de cada categoría
````

## Parte 5 · El data frame: una base de datos

Un vector guarda **una** variable. Pero una base de datos real tiene muchas variables sobre las
mismas personas: edad, comuna, transporte, horas de sueño...

Ese objeto —una tabla con **filas** (casos/personas) y **columnas** (variables)— se llama
**data frame**, y es la estructura que vamos a usar durante todo el semestre.

Piensa en una planilla de Excel: filas y columnas. Es exactamente lo mismo, pero manipulable con
código (y por lo tanto reproducible y auditable, que es la razón por la que no usamos Excel).

Construyamos uno chico a mano, con `data.frame()`:

````r
curso_mini <- data.frame(
  nombre     = c("Ana", "Beto", "Cami", "Dani", "Eli"),
  edad       = c(18, 19, 18, 21, 20),
  comuna     = c("Ñuñoa", "Maipú", "Ñuñoa", "La Florida", "Providencia"),
  horas_redes = c(2.5, 4.0, 3.5, 1.0, 5.5)
)

curso_mini
````

Para mirar **una sola columna** usamos el signo `$`:

````r
curso_mini$edad            # la columna edad, que es un vector

mean(curso_mini$edad)      # y por lo tanto podemos calcular su promedio
````

> 🔑 **La regla de oro:** `base$columna` se lee *"la columna `columna` de la base `base`"*.

````r
barplot(table(curso$transporte),
        main = "¿Cómo llega el curso a la universidad?",
        ylab = "Número de estudiantes",
        col  = "darkorange")
````

## Parte 9 · Filtrar: mirar solo una parte de la base

Muchas veces la pregunta no es sobre todo el curso, sino sobre un subgrupo:
*¿cuánto viajan los que vienen en micro?*

Para eso usamos los **corchetes** `[ ]`, que se leen como *"quédate solo con las filas donde..."*.
La estructura es `base[condición, ]` — la coma al final significa *"y déjame todas las columnas"*.

Los operadores de comparación son:

| Operador | Significa |
|---|---|
| `==` | es igual a (¡**dos** signos igual!) |
| `!=` | es distinto de |
| `>` `<` | mayor que, menor que |
| `>=` `<=` | mayor o igual, menor o igual |

````r
# Solo quienes vienen en micro
micreros <- curso[curso$transporte == "Micro", ]

nrow(micreros)                      # ¿cuántos son?
mean(micreros$minutos_viaje_u)      # ¿cuánto viajan en promedio?
````

Comparemos ese promedio con el del curso completo:

````r
mean(curso$minutos_viaje_u)         # promedio de todo el curso
````

Otro ejemplo, ahora con una condición numérica:

````r
# Quienes duermen menos de 6 horas
poco_sueno <- curso[curso$horas_sueno < 6, ]

nrow(poco_sueno)
table(poco_sueno$toma_cafe)
````
