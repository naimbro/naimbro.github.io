**Descripción y Visualización de Datos · UAI 2026 · Clase 13**

Hoy graban un video-podcast de tres minutos. Mientras conversan, el gráfico se arma en pantalla
por etapas: primero el marco, después la línea, después la referencia y al final el dato. Cada
etapa es un PNG, y cada PNG es una lámina.

| | Qué hace |
|---|---|
| `p <- ggplot(...)` | guarda el gráfico con un nombre, sin dibujarlo |
| `p <- p + geom_line()` | le suma una capa al gráfico guardado |
| `scale_y_continuous(limits = c(0, 40))` | fija el eje: no se mueve entre etapas |
| `geom_hline(yintercept = 19)` | una línea horizontal de referencia |
| `annotate("text", x = ..., y = ..., label = "...")` | un texto en un punto del gráfico |
| `ggsave("etapa1.png", p, width = 8, height = 5)` | guarda esa etapa como PNG |

## Parte 0 · Punto de partida

```r
library(dplyr)
library(ggplot2)
options(repr.plot.width = 9, repr.plot.height = 5)

problemas <- read.csv("https://raw.githubusercontent.com/naimbro/naimbro.github.io/main/materiales/2026_descripcion_visualizacion_datos/datos/cep_problemas_por_anio.csv")

delinc <- problemas %>%
  filter(problema == "Delincuencia, asaltos y robos")
```

## Parte 1 · `p <-`: el gráfico con nombre

Si el gráfico se guarda en `p`, no se dibuja: se dibuja cuando escriben `p`. La primera etapa es
el marco, con el eje fijo de 0 a 40. Así el eje no salta cuando se suman capas.

```r
p <- ggplot(delinc, aes(x = anio, y = porcentaje)) +
  scale_y_continuous(limits = c(0, 40))
p
ggsave("etapa1.png", p, width = 8, height = 5)
```

## Parte 3 · `geom_hline()`: la línea de referencia

Una línea horizontal le da al que mira un punto de comparación. Aquí, el punto de partida: 19%
en 1994.

```r
p <- p + geom_hline(yintercept = 19, linetype = "dashed")
p
ggsave("etapa3.png", p, width = 8, height = 5)
```

## Parte 4 · `annotate()`: el dato que dice el conductor

`annotate("text", ...)` escribe un texto en el punto `x`, `y` del gráfico. Es el número que el
conductor dice en voz alta.

```r
p <- p + annotate("text", x = 2019, y = 10,
                  label = "2019: 14%")
p
ggsave("etapa4.png", p, width = 8, height = 5)
```

## Lo que aprendiste hoy

| Para qué | Cómo se escribe |
|---|---|
| Guardar el gráfico con un nombre | `p <- ggplot(...)` |
| Dibujarlo | `p` |
| Sumarle una capa | `p <- p + geom_line()` |
| Que el eje no salte entre etapas | `scale_y_continuous(limits = c(0, 40))` |
| Una línea de referencia | `geom_hline(yintercept = 19)` |
| Un texto en el gráfico | `annotate("text", x = 2019, y = 10, label = "2019: 14%")` |
| Una etapa, un PNG | `ggsave("etapa1.png", p, width = 8, height = 5)` |
