## Parte 1 · R en tres minutos

Sólo necesitas tres ideas para empezar.

### Idea 1: R calcula

Ejecuta la celda (recuerda: **Shift + Enter**).

### Idea 2: R guarda cosas con nombre

La flecha `<-` significa *"guarda esto adentro de"*. A lo guardado le llamamos **objeto**.

```
mi_edad  <-  19
   ↑         ↑
 nombre    valor
```

````r
mi_edad <- 19        # guardamos el número 19 con el nombre "mi_edad"
mi_edad + 1          # y ahora podemos usarlo
````

Reglas rápidas para los nombres: sin espacios (usa `guion_bajo`), no empiezan con número, y R
**distingue mayúsculas de minúsculas** (`Edad` ≠ `edad`).

### Idea 3: R tiene funciones

Una **función** es una orden con un nombre, y recibe lo que necesita entre paréntesis:
`mean(algo)` calcula el promedio de *algo*.

Y algunas cosas van entre comillas y otras no. Los **números** van pelados; el **texto** siempre
entre comillas:

````r
altura <- 1.72          # número: sin comillas
comuna <- "Ñuñoa"       # texto: CON comillas

class(altura)           # class() nos dice qué tipo de dato es
class(comuna)
````

## Parte 3 · Ponerle nombres decentes a las columnas

Fíjate en los nombres de las columnas: son las preguntas completas del formulario, convertidas por
R en algo ilegible.

````r
names(respuestas)
````

Trabajar con `X.Cuál.es.tu.estatura.en.centímetros..Ingresa.sólo.el.número.` es insoportable. Así
que **renombramos**: le pasamos a `names()` un vector con nombres cortos.

Un **vector** se construye con `c()` (de *combinar*), separando los elementos con comas. Es la
segunda estructura básica de R, después del data frame.

> ⚠️ El orden importa: los nombres nuevos se asignan a las columnas **en el orden en que están**.
> Y tiene que haber exactamente uno por columna (aquí, 17).

````r
names(respuestas) <- c("marca_temporal", "edad", "estatura", "hermanos", "comuna",
                       "minutos_viaje", "transporte", "horas_sueno", "horas_redes",
                       "sistema_operativo", "tazas_cafe", "experiencia", "dominio",
                       "op_grafico_miente", "op_interes_programar",
                       "op_datos_chile", "op_hablar_publico")

names(respuestas)     # revisamos que quedó bien
````

Ahora sí. Para mirar **una columna** usamos el signo `$`:

> 🔑 **La regla de oro:** `base$columna` se lee *"la columna `columna` de la base `base`"*.
> Si entiendes el `$`, entiendes el 80% de lo que haremos las próximas semanas.

````r
respuestas$edad          # la columna edad completa
````

Y `str()` da la radiografía de la base: cada variable, su tipo y sus primeros valores.

````r
str(respuestas)
````

Fíjate en la etiqueta que aparece al lado de cada variable:

- `int` o `num` → **numérica**: tiene sentido preguntarle un promedio.
- `chr` → **texto** (categórica): tiene sentido preguntarle cuántos hay de cada tipo.

Esa distinción no es un detalle técnico: **define qué preguntas puedes hacerle a cada variable**.
Preguntar el promedio de las comunas no significa nada.

## Parte 4 · Describir el curso

### Variables numéricas: promedio, mediana, dispersión

Estas son las medidas que vimos la clase pasada, ahora calculadas con una línea cada una.

````r
mean(respuestas$edad)             # media (promedio)
median(respuestas$minutos_viaje)  # mediana (el valor del medio)
sd(respuestas$horas_sueno)        # desviación estándar (cuánto se dispersan los datos)
````

`summary()` entrega todo de una vez: mínimo, cuartiles, mediana, media y máximo.

````r
summary(respuestas$minutos_viaje)
````

Y `round()` deja el resultado con los decimales que quieras. Reportar `38.4` minutos informa;
reportar `38.42105263157895` sólo hace ruido.

````r
round(mean(respuestas$minutos_viaje), 1)
````

### Variables categóricas: frecuencias

Aquí la pregunta no es el promedio, sino **cuántos hay en cada categoría**. Eso lo hace `table()`.

````r
table(respuestas$transporte)
````

Los conteos absolutos se comparan mal. Casi siempre queremos **porcentajes**, y para eso está
`prop.table()`:

````r
round(prop.table(table(respuestas$transporte)) * 100, 1)
````

Dos que interesan especialmente para este curso — cuánta experiencia previa hay en la sala, y qué
dominios les llaman la atención para el proyecto:

````r
table(respuestas$experiencia)
````

````r
sort(table(respuestas$dominio), decreasing = TRUE)   # ordenado de mayor a menor
````

### La moda

R no trae una función para la moda (el valor más frecuente), pero se arma en un paso a partir de
la tabla de frecuencias:

````r
names(which.max(table(respuestas$comuna)))   # la comuna más frecuente
````
