## Parte 4 · Describir el curso

### Variables numéricas: promedio, mediana, dispersión

Estas son las medidas que vimos la clase pasada, ahora calculadas con una línea cada una.

````r
mean(respuestas$edad)             # media (promedio)
median(respuestas$minutos_viaje)  # mediana (el valor del medio)
sd(respuestas$horas_sueno)        # desviación estándar (cuánto se dispersan los datos)
````

`summary()` entrega todo de una vez: mínimo, cuartiles, mediana, media y máximo.

````r
summary(respuestas$minutos_viaje)
````

Y `round()` deja el resultado con los decimales que quieras. Reportar `38.4` minutos informa;
reportar `38.42105263157895` sólo hace ruido.

````r
round(mean(respuestas$minutos_viaje), 1)
````

### Variables categóricas: frecuencias

Aquí la pregunta no es el promedio, sino **cuántos hay en cada categoría**. Eso lo hace `table()`.

````r
table(respuestas$transporte)
````

Los conteos absolutos se comparan mal. Casi siempre queremos **porcentajes**, y para eso está
`prop.table()`:

````r
round(prop.table(table(respuestas$transporte)) * 100, 1)
````

Dos que interesan especialmente para este curso — cuánta experiencia previa hay en la sala, y qué
dominios les llaman la atención para el proyecto:

````r
table(respuestas$experiencia)
````

````r
sort(table(respuestas$dominio), decreasing = TRUE)   # ordenado de mayor a menor
````

### La moda

R no trae una función para la moda (el valor más frecuente), pero se arma en un paso a partir de
la tabla de frecuencias:

````r
names(which.max(table(respuestas$comuna)))   # la comuna más frecuente
````

## Parte 5 · Un dato sucio: la estatura

Intenta calcular la estatura promedio:

````r
mean(respuestas$estatura)
````

### Mira bien lo que pasó

R **no** se cayó con un error. Devolvió `NA` (dato faltante) y una advertencia:
`argument is not numeric or logical`. Traducido: *"me pediste el promedio de algo que no es un
número, así que te devuelvo un vacío"*.

> 🧠 **Primera lección importante del día:** un código que corre sin errores no es un código
> correcto. Hay que mirar el resultado.

Miremos qué contiene realmente esa columna:

````r
head(respuestas$estatura, 20)
````

Ahí está el problema: **cada persona escribió su estatura como quiso**. Unos pusieron `172`, otros
`1,72`, otros `1.72`, alguno escribió `172 cm`. Como hay letras y comas, R guardó toda la columna
como texto.

Esto no es un accidente de nuestra encuesta: **es lo normal**. Cuando le pides un número a una
persona en un formulario abierto, recibes formatos mezclados. Gran parte del trabajo real con
datos consiste exactamente en esto, y por eso el Sprint 2 se llama "prototipo de datos".

### La receta de limpieza

Tres pasos, de adentro hacia afuera:

1. `gsub("[^0-9.,]", "", x)` → **borra todo lo que no sea** un dígito, un punto o una coma
   (adiós al `cm`).
2. `gsub(",", ".", x)` → cambia la **coma decimal por punto**, que es lo que R entiende.
3. `as.numeric(x)` → **convierte** el texto resultante en número.

````r
estatura_limpia <- respuestas$estatura              # partimos de la columna original
estatura_limpia <- gsub("[^0-9.,]", "", estatura_limpia)   # 1. sacar letras y espacios
estatura_limpia <- gsub(",", ".", estatura_limpia)         # 2. coma  ->  punto
estatura_limpia <- as.numeric(estatura_limpia)             # 3. texto ->  número

summary(estatura_limpia)
````

Ya es numérica, pero mira el mínimo y el máximo: conviven valores como `1.55` con valores como
`185`. **Son la misma estatura en unidades distintas**: unos contestaron en metros y otros en
centímetros.

La regla es simple: si el valor es menor que 3, está en metros → lo multiplicamos por 100.

Para eso usamos los **corchetes** `[ ]`, que seleccionan sólo algunos elementos de un vector, y
`which()`, que entrega **las posiciones donde se cumple una condición**:

````r
# "de estatura_limpia, los elementos menores a 3, multiplícalos por 100"
estatura_limpia[which(estatura_limpia < 3)] <- estatura_limpia[which(estatura_limpia < 3)] * 100

summary(estatura_limpia)
````

Ahora sí: todos los valores están en centímetros y el promedio significa algo.

Guardemos la variable limpia **como una columna nueva** de la base. Nunca borres el dato original:
la trazabilidad —poder mostrar de dónde salió cada número— es uno de los criterios transversales
del curso.

````r
respuestas$estatura_cm <- estatura_limpia    # columna nueva, el original queda intacto

round(mean(respuestas$estatura_cm), 1)       # estatura promedio del curso, en cm
````

## Parte 6 · Los primeros gráficos

Una tabla de 40 números cuesta leerla; un gráfico se entiende en dos segundos. Esa es, literalmente,
la razón de ser de este curso.

Por ahora usamos las funciones gráficas básicas de R. En la clase 8 pasaremos a `ggplot2`.

### Histograma: cómo se distribuye una variable numérica

````r
hist(respuestas$estatura_cm)
````

Funciona, pero está feo y —más importante— **no se entiende solo**: el título es el nombre de una
variable interna y el eje X no dice unidades. Un gráfico siempre debe poder leerse sin que nadie
lo explique.

````r
hist(respuestas$estatura_cm,
     main = "Distribución de la estatura del curso",   # título
     xlab = "Estatura (cm)",                           # nombre del eje X
     ylab = "Número de estudiantes",                   # nombre del eje Y
     col  = "steelblue")                               # color de las barras
````

> 🎨 Compara los dos gráficos: el cálculo es idéntico, lo único que cambió fue el cuidado en la
> comunicación. Esa diferencia es la que evaluaremos en los sprints.

### Gráfico de barras: cómo se reparte una variable categórica

`barplot()` no recibe la variable, sino su **tabla de frecuencias**. Por eso va con `table()` adentro:

````r
barplot(table(respuestas$transporte),
        main = "¿Cómo llega el curso a la universidad?",
        ylab = "Número de estudiantes",
        col  = "darkorange")
````

### Dispersión: ¿se relacionan dos variables numéricas?

````r
plot(respuestas$horas_redes, respuestas$horas_sueno,
     main = "Redes sociales y horas de sueño",
     xlab = "Horas diarias en redes sociales",
     ylab = "Horas que durmió anoche",
     pch  = 19,               # tipo de punto (19 = círculo relleno)
     col  = "darkred")
````

> ⚠️ **Advertencia que repetiremos todo el semestre:** aunque la nube se vea inclinada, eso **no**
> prueba que las redes quiten sueño. Podría ser al revés, o podría haber un tercer factor detrás de
> ambas. Este curso es de *descripción*: mostramos patrones con honestidad y dejamos explícitos los
> límites de lo que podemos afirmar.

### Boxplot: comparar una variable numérica entre grupos

La virgulilla `~` se lee **"según"**: `minutos_viaje ~ transporte` es *"los minutos de viaje según
el medio de transporte"*.

````r
boxplot(minutos_viaje ~ transporte,
        data = respuestas,
        main = "Tiempo de viaje según medio de transporte",
        xlab = "Medio de transporte",
        ylab = "Minutos",
        col  = "lightgreen")
````
