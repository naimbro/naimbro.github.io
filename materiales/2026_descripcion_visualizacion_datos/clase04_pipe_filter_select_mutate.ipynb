Seis personas. Y fíjate en el detalle que más errores causa en todo el curso:

> **`==` son DOS iguales.** Uno solo (`=`) significa *"guarda esto acá"*. Dos
> significan *"¿es igual a?"*. Son cosas distintas y R no perdona la confusión.

El texto que buscas va **entre comillas** y tiene que estar escrito exactamente
como aparece en la base: `"Metro"` con mayúscula funciona, `"metro"` no.

Otra cosa importante: `filter()` **no toca `curso`**. Devuelve una tabla nueva y
deja la original intacta. Compruébalo:

````r
metro <- filter(curso, transporte == "Metro")

nrow(metro)
nrow(curso)
````

Además de `==`, dentro de `filter()` puedes usar:

| Operador | Significa | Ejemplo |
|---|---|---|
| `==` | es igual a | `transporte == "Metro"` |
| `!=` | es distinto de | `transporte != "Auto"` |
| `>` &nbsp;`<` | mayor / menor que | `edad > 19` |
| `>=` &nbsp;`<=` | mayor o igual / menor o igual | `tazas_cafe >= 2` |

## Parte 4 · `%>%` · el pegamento

Ahora la pregunta de verdad: *"¿de qué comunas vienen los que llegan en metro, y
cuánto se demoran?"*

Eso son **dos pasos**: primero filtrar filas, después elegir columnas. Se puede
hacer guardando un resultado intermedio:

````r
paso_1 <- filter(curso, transporte == "Metro")
paso_2 <- select(paso_1, comuna, minutos_viaje)

paso_2
````

Funciona, pero tuviste que inventar dos nombres (`paso_1`, `paso_2`) que no le
importan a nadie. Con tres o cuatro pasos eso se vuelve insoportable.

El operador `%>%` —se llama **pipe**, "tubo"— resuelve exactamente eso:

````r
curso %>%
  filter(transporte == "Metro") %>%
  select(comuna, minutos_viaje)
````

Mismo resultado, sin nombres intermedios. **Léelo en voz alta y vas a ver que se
entiende solo:**

> toma `curso`, **y luego** quédate con los que van en metro, **y luego**
> muéstrame comuna y minutos de viaje.

La regla, en una línea: **`%>%` agarra lo que hay a su izquierda y se lo entrega
como primer argumento a la función de su derecha.** Por eso adentro de
`filter()` ya no escribes `curso`: el pipe se lo pasó.

Dos cosas prácticas:

- El `%>%` va **al final** de la línea, nunca al principio de la siguiente.
- Hay un atajo de teclado para escribirlo: **`Ctrl` + `Shift` + `M`**.

## Parte 5 · `mutate()` · agregar una columna nueva

`filter()` y `select()` sacan cosas. `mutate()` es la única de las tres que
**agrega**: inventa una columna a partir de las que ya existen.

````r
curso %>%
  mutate(tazas_semana = tazas_cafe * 7) %>%
  select(tazas_cafe, tazas_semana) %>%
  head()
````

La forma es siempre la misma:

```
mutate(nombre_nuevo = fórmula)
```

A la izquierda del `=` el nombre que **tú** eliges (uno solo, sin comillas, sin
espacios). A la derecha, el cálculo. La columna nueva queda al final de la tabla;
el `select()` de arriba está sólo para poder verla al lado de su origen.

Pregunta razonable: *"¿cuántas personas se demoran más de una hora en llegar?"*.
Se escribe sola:

````r
curso %>%
  filter(minutos_viaje > 60) %>%
  count(minutos_viaje)
````

Diez personas. **Y está mal.**

Mira la lista de valores que sobrevivieron: 70, 80, 85, 90. ¿Y los que pusieron
100, 130, 150 y 180? Son los que **más** se demoran de todo el curso, y el filtro
los dejó fuera.

La razón es de la clase pasada: alguien escribió `10 min` en esa columna, así que
`minutos_viaje` no es un número, es **texto**. Y R, al comparar texto, no mide:
**deletrea**. Compara `"130"` con `"60"` letra por letra, ve que `1` viene antes
que `6`, y concluye que 130 es menor que 60.

> Lo peligroso no es el resultado: es que **no hubo ningún error ni ninguna
> advertencia**. La respuesta salió limpia, con cara de correcta.

### Lo que aprendiste hoy

| Para qué | Cómo se escribe |
|---|---|
| Quedarse con algunas filas | `filter(curso, transporte == "Metro")` |
| Quedarse con algunas columnas | `select(curso, edad, comuna)` |
| Sacar una columna | `select(curso, -estatura)` |
| Agregar una columna nueva | `mutate(tazas_semana = tazas_cafe * 7)` |
| Una columna de `TRUE`/`FALSE` | `mutate(es_apple = sistema_operativo == "iOS (Apple)")` |
| Encadenar pasos | `curso %>% filter(...) %>% select(...)` |
| Escribir el pipe rápido | `Ctrl` + `Shift` + `M` |
| Guardar el resultado | `curso <- curso %>% mutate(...)` |

### Las tres ideas

1. **`==` pregunta, `=` guarda.** Es el error más común del semestre y no da un
   mensaje que ayude.
2. **El pipe se lee como una receta.** `%>%` es simplemente "y luego". Si tu
   cadena no se puede leer en voz alta, probablemente está mal armada.
3. **Un filtro puede mentir sin avisar.** `minutos_viaje > 60` devolvió diez
   personas y dejó fuera justo a las cuatro que más viajan, sin un solo error en
   pantalla.
