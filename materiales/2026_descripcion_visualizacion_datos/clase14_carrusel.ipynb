# Cuaderno 14 — El carrusel: `slice_max()` y `geom_text()`

**Descripción y Visualización de Datos · UAI 2026 · Clase 14**

Hoy arman un carrusel: seis láminas cuadradas, cada una entendible en tres segundos. En tres
segundos nadie lee un eje: la lámina dice una sola cosa, con el número escrito y el hallazgo
en rosa.

| | Qué hace |
|---|---|
| `slice_max(porcentaje, n = 5)` | se queda con las cinco filas con el valor más alto |
| `+ geom_text(aes(label = ...))` | escribe el número al lado de cada barra |
| `mutate(destacado = ...)` + `scale_fill_manual()` | pinta en rosa la barra del hallazgo |
| `ggsave("lamina.png", width = 4, height = 4)` | guarda la lámina cuadrada |

Los datos son los de siempre: la serie de problemas de la CEP 1994–2026 (sin 2020).

## Parte 0 · Punto de partida

```r
library(dplyr)
library(ggplot2)
options(repr.plot.width = 7, repr.plot.height = 5)

problemas <- read.csv("https://raw.githubusercontent.com/naimbro/naimbro.github.io/main/materiales/2026_descripcion_visualizacion_datos/datos/cep_problemas_por_anio.csv")
```

## Parte 1 · `slice_max()`: quedarse con los primeros

En la clase 9 el ranking se hacía con `arrange(desc(porcentaje)) %>% head(8)`. `slice_max()`
hace lo mismo en un paso: ordena por la variable y se queda con las `n` primeras filas.

```r
top5 <- problemas %>%
  filter(anio == 2026) %>%
  slice_max(porcentaje, n = 5)

ggplot(top5, aes(x = porcentaje,
                 y = reorder(problema, porcentaje))) +
  geom_col()
```

## Parte 2 · `geom_text()`: el número en la barra

`geom_text()` escribe un texto en cada fila. Lo que se escribe va en `aes(label = ...)`:
aquí, el porcentaje redondeado con `%` al final. `nudge_x` lo corre un poco a la derecha,
para que no quede encima del final de la barra.

```r
ggplot(top5, aes(x = porcentaje,
                 y = reorder(problema, porcentaje))) +
  geom_col() +
  geom_text(aes(label = paste0(round(porcentaje), "%")),
            nudge_x = 2)
```

## Parte 3 · Una barra en rosa: el hallazgo

La clase 9 pintó las barras con `scale_fill_manual()`. Hoy se pinta sólo una: `mutate()` crea
una columna que es `TRUE` en el hallazgo y `FALSE` en el resto, y `scale_fill_manual()` le da
gris al `FALSE` y rosa al `TRUE`. `guide = "none"` saca la leyenda, que en una lámina sobra.

```r
top5 <- top5 %>%
  mutate(destacado = problema ==
           "Delincuencia, asaltos y robos")

ggplot(top5, aes(x = porcentaje,
                 y = reorder(problema, porcentaje),
                 fill = destacado)) +
  geom_col() +
  geom_text(aes(label = paste0(round(porcentaje), "%")),
            nudge_x = 2) +
  scale_fill_manual(values = c("grey75", "#C4878C"),
                    guide = "none")
```

## Parte 4 · La lámina cuadrada

Un carrusel se mira en el celular: las láminas son cuadradas. `ggsave()` con
`width = 4, height = 4` la guarda así. En un cuadrado los nombres largos comen espacio, así
que `scale_x_continuous(limits = c(0, 40))` deja lugar para el número de la barra más larga.
El título es el hallazgo, como siempre.

```r
ggplot(top5, aes(x = porcentaje,
                 y = reorder(problema, porcentaje),
                 fill = destacado)) +
  geom_col() +
  geom_text(aes(label = paste0(round(porcentaje), "%")),
            nudge_x = 4) +
  scale_fill_manual(values = c("grey75", "#C4878C"),
                    guide = "none") +
  scale_x_continuous(limits = c(0, 40)) +
  labs(title = "1 de cada 3: la delincuencia",
       x = NULL, y = NULL,
       caption = "Fuente: CEP 2026")

ggsave("lamina2.png", width = 4, height = 4)
```

| Para qué | Cómo se escribe |
|---|---|
| Quedarse con los primeros | `slice_max(variable, n = 5)` |
| El número en la barra | `+ geom_text(aes(label = paste0(round(variable), "%")), nudge_x = 2)` |
