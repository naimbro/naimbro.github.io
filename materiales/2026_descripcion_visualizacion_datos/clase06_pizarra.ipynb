La receta, por si hay que volver a ella:

```r
cep %>%
  mutate(delinc = problema_1 == "Delincuencia, asaltos y robos") %>%
  group_by(anio) %>%
  summarise(personas = n(),
            pct = mean(delinc) * 100) %>%
  arrange(desc(pct))
```
