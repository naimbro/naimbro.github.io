## 1 · ¿Qué es Google Colab?

**Colab** (abreviación de *Colaboratory*) es una herramienta gratuita de Google para escribir y
ejecutar código **desde el navegador**. Tres cosas que conviene tener claras desde el principio:

- **No instalas nada.** El código no corre en tu computador: corre en un computador de Google al
  que te conectas por internet. Un Chromebook viejo y un notebook caro funcionan igual.
- **Se guarda en tu Google Drive**, como un documento cualquiera. Y se comparte igual que un
  Google Doc: con un link.
- **Mezcla texto y código.** Un cuaderno no es sólo un programa: es un documento donde conviven la
  explicación, el código, los resultados y los gráficos. Por eso es ideal para documentar un
  análisis... y para hacer la bitácora de su proyecto.

> 💡 Esto último es más importante de lo que parece. En este curso no basta con que el código
> funcione: tiene que quedar claro **por qué** hiciste lo que hiciste. Un cuaderno es exactamente
> el formato para eso.

## 3 · Las celdas: los ladrillos del cuaderno

Un cuaderno está hecho de **celdas**, y sólo hay de dos tipos:

| Tipo | Qué contiene | Cómo se ve |
|---|---|---|
| **Celda de texto** | Explicaciones, títulos, comentarios para humanos | Como esta que estás leyendo |
| **Celda de código** | Instrucciones que el computador ejecuta | Fondo gris, con un ▶ a la izquierda |

Abajo hay una celda de código. **Ejecútala** de cualquiera de estas tres formas:

- clic en el botón **▶** que aparece a su izquierda,
- **Shift + Enter** → ejecuta y salta a la celda siguiente,
- **Ctrl + Enter** → ejecuta y se queda donde está.

La primera vez demora unos segundos: Google está "prendiendo" el computador que te asignó. Vas a
ver un círculo girando y, cuando termine, un ✓ verde.

````r
print("Hola, soy R y estoy corriendo en un computador de Google")
````

¿Apareció el saludo justo debajo de la celda? Entonces ya ejecutaste tu primer código. 🎉

Fíjate en dos detalles:

- El resultado aparece **abajo de la celda**, no en otra ventana.
- A la izquierda apareció un número —`[1]`, `[2]`...— que indica **el orden** en que fuiste
  ejecutando las celdas. Es útil para saber qué ejecutaste y qué no.

Los `[1]` que a veces aparecen *dentro* del resultado son otra cosa (una peculiaridad de R para
numerar los valores que muestra). Por ahora, ignóralos.

### Los comentarios

Dentro de una celda de código, todo lo que va después de un `#` es un **comentario**: R lo ignora
por completo. Sirve para dejar notas.

````r
# Esto es un comentario: no hace nada, es una nota para quien lea el código.
2 + 2       # también puede ir al final de una línea
````

## 6 · 🔀 R y Python: dónde se cambia

Colab sirve para varios lenguajes de programación. Los dos principales son **Python** (el que
viene por defecto) y **R** (el que usaremos en este curso).

**Un cuaderno usa un solo lenguaje a la vez.** No se mezclan: o es un cuaderno de R, o es uno de
Python. Este ya viene configurado en R.

### Dónde se cambia

> **Entorno de ejecución → Cambiar tipo de entorno de ejecución**

Se abre una ventanita con un menú desplegable donde eliges **R** o **Python 3**. (En esa misma
ventana está la opción de acelerador por hardware — GPU, TPU — que no necesitaremos.)

> ⚠️ **Ojo:** cambiar de lenguaje **reinicia el cuaderno y borra todo lo calculado**. El texto y el
> código escrito no se pierden, pero los resultados sí: hay que volver a ejecutar todo.

### Cómo saber en cuál estás

La forma más rápida es preguntárselo al cuaderno. Ejecuta esto:

````r
R.version.string
````

Si respondió algo como `'R version 4.4.1'`, estás en R. **Si en cambio aparece un error rojo que
dice `NameError` o `SyntaxError`, estás en Python**: anda al menú y cámbialo a R.

### ¿Y por qué R y no Python?

Los dos son excelentes y se parecen bastante. Elegimos R porque nació dentro de la estadística y
las ciencias sociales: leer una base, describirla y graficarla requiere menos código y menos
conceptos previos. Además `ggplot2` —la librería de gráficos que veremos en la clase 8— es de lo
mejor que existe para visualización.

Como referencia, así se ven las mismas tres ideas en cada lenguaje:

| Idea | En R | En Python |
|---|---|---|
| Comentario | `# nota` | `# nota` |
| Mostrar algo | `print("hola")` | `print("hola")` |
| Guardar un valor | `x <- 10` | `x = 10` |
| Promedio de una columna | `mean(datos$edad)` | `datos["edad"].mean()` |

Se parecen. Si aprendes uno, el otro te costará mucho menos.

## 7 · El entorno de ejecución (y por qué a veces "todo se rompe")

El **entorno de ejecución** es el computador de Google que te prestaron. Tres cosas que hay que
saber de él:

1. **Tiene memoria.** Lo que calculas en una celda queda disponible para las siguientes.
2. **El orden importa**, y es el orden en que *tú ejecutas*, no el orden en que las celdas están
   escritas en la página.
3. **Se desconecta solo** después de un rato de inactividad (o al cerrar la pestaña por mucho
   tiempo). Cuando eso pasa, pierdes lo calculado —no lo escrito— y hay que volver a ejecutar.

Veamos la memoria en acción. Ejecuta esta celda:

````r
mi_numero <- 10     # guardamos un valor con el nombre "mi_numero"
````

No mostró nada, y está bien: guardar un valor no imprime nada. Ahora ejecuta la siguiente celda,
que usa lo que quedó guardado:

````r
mi_numero * 3
````

Funcionó porque la celda anterior se ejecutó primero. **Si hubieras ejecutado esta celda sola, en
un cuaderno recién abierto, habrías recibido el error `object 'mi_numero' not found`** — que no
significa que el código esté malo, sino que ese objeto todavía no existe en la memoria.

### El botón que arregla el 90% de los problemas

Cuando algo deje de funcionar sin razón aparente:

> **Entorno de ejecución → Reiniciar y ejecutar todo**

Eso borra la memoria y vuelve a ejecutar el cuaderno completo, de arriba hacia abajo, en orden.
Es el equivalente a apagar y prender.

## 8 · 🤖 Dónde se apaga la IA del cuaderno

Colab viene con un asistente de IA (**Gemini**) activado por defecto: sugiere código mientras
escribes, agrega un botón de *"Explicar error"* cuando algo falla y abre un panel de chat desde el
ícono ✦ del pie del cuaderno.

Hay momentos en que conviene apagarlo —por ejemplo, mientras estás aprendiendo a escribir el
código tú mismo, o si se pide expresamente en una evaluación—. Conviene que sepas dónde está el
interruptor.

### Dónde se apaga

> **Herramientas → Configuración → Asistencia de IA**
> (si tu Colab está en inglés: *Tools → Settings → AI Assistance*)

En esa pestaña desmarca las casillas de las funciones generativas —la que habilita el asistente y
la de sugerencias automáticas de código— y, si prefieres que ni siquiera aparezcan los botones,
marca **"Ocultar funciones de IA generativa"**. Después, **Guardar**.

Si no ves la pestaña *Asistencia de IA*, busca las sugerencias automáticas en la pestaña
**Editor**, dentro de la misma ventana de configuración.

> ⚠️ Es una configuración **de tu cuenta de Colab, no de este cuaderno**: se aplica a todos tus
> cuadernos, y se revierte con los mismos pasos cuando quieras volver a usar la IA.

## 9 · Otras cosas útiles de la interfaz

- **Panel lateral izquierdo** (los íconos verticales):
  - 📑 *Índice*: la lista de títulos del cuaderno, para navegar rápido.
  - 📁 *Archivos*: los archivos que hay en el computador prestado. Se pueden **subir arrastrando**
    (útil si tienes un `.csv` en tu computador). ⚠️ Se borran al desconectarse el entorno.
- **Botón Compartir** (arriba a la derecha): funciona igual que en Google Docs. Así entregarás
  trabajos y así trabajarán en grupo sobre el mismo cuaderno.
- **Atajos que vale la pena memorizar:**

| Atajo | Qué hace |
|---|---|
| `Shift + Enter` | Ejecuta la celda y pasa a la siguiente |
| `Ctrl + Enter` | Ejecuta la celda y se queda ahí |
| `Ctrl + M` y luego `B` | Inserta una celda de código abajo |
| `Ctrl + S` | Guarda |
| `Ctrl + Z` | Deshace (¡también deshace borrar una celda!) |
