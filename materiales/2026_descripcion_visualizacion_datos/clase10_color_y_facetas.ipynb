# Cuaderno 10 — Tres variables: `color =` y `facet_wrap()`

**Descripción y Visualización de Datos · UAI 2026 · Clase 10**

Hasta ahora sus gráficos tenían dos variables: un eje x y un eje y. Hoy suman una tercera,
de dos formas:

| | Qué hace | Cuándo |
|---|---|---|
| `color = problema` dentro de `aes()` | una línea de cada color, todas en el mismo plano | pocos grupos que se comparan entre sí |
| `+ facet_wrap(~ problema)` | un panel por grupo, todos con el mismo eje | muchos grupos, o cuando importa la forma de cada uno |

Los datos son los de la clase 6: qué porcentaje de la CEP nombra cada problema como el
principal del país, de 1994 a 2026 (sin 2020).

## Parte 3 — `facet_wrap()`: un panel por grupo

Con ocho problemas, ocho colores no se leen. `facet_wrap(~ problema)` hace un panel por
problema. Ojo con el `~` antes del nombre.

````r
ocho <- problemas %>%
  filter(problema %in% c("Delincuencia, asaltos y robos",
                         "Salud", "Pobreza", "Empleo",
                         "Educación", "Sueldos",
                         "Corrupción", "Vivienda"))

ggplot(ocho, aes(x = anio, y = porcentaje)) +
  geom_line() +
  facet_wrap(~ problema)
````

## Lo que aprendiste hoy

| Para qué | Cómo se escribe |
|---|---|
| Una línea por grupo, en el mismo plano | `aes(x = ..., y = ..., color = grupo)` |
| Un panel por grupo | `+ facet_wrap(~ grupo)` |
| Cuántos paneles por fila | `facet_wrap(~ grupo, ncol = 2)` |

**La regla:** pocos grupos que se comparan → `color =`. Muchos grupos, o la forma de cada
uno → `facet_wrap()`. Si para leer el gráfico hay que ir y volver a la leyenda, pasa a
facetas.
